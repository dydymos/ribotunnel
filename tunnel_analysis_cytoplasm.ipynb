```python
###############
## Reference ##
###############
# Run reference path
# Defined as a dictionary to match our updated formatting
reference_org = {'ecun': 'E.cuniculi'}

path = "./data_new/eukaryota/" 

ref_dict = get_dictionary(reference_org)
```

```python
# Grid for calculations
grid_resolution = 1.0  # Grid resolution in Angstroms

# getting grid for each trajectory and finding the common one with the use of get_grid.py script
# When grid is ready we can calculate occupancy map for each trajectory using -> occupancy_map.py script

grid_file = "./data/grid_data.dat"
grid_dim_min, grid_dim_max = read_grid(grid_file)
```

```python
# Occupancy path
ref_id = list(reference_org.keys())

# Read occupancy max path
read_occupancy_path(path, ref_dict, ref_id)

# Calculate the splines
# s=60 controls smoothing, NN=25 is the tunnel exit cutoff index
calculate_spline(ref_dict, ref_id, s=60, NN=25, short=False)

# write occ path avg, its interpolation and spline into PDB files:
write_paths(path,ref_dict,ref_id,NN=25)

# Reference tunnel path
# Using the 60-length nascent chain spline from E. cuniculi as the baseline
ref_tunnel_path = np.array(ref_dict[ref_id[0]]['spline path avg']['60'])
```

```python
#######################################
# Selection of the group of organisms #
#######################################
theme = "micro"

# Env setup: Map themes to their respective relative data paths
theme_paths = {
    "bacteria": "./data_new/bacteria/",
    "eukaryota": "./data_new/eukaryota/",
    "eukaryota_core": "./data_new/eukaryota/",
    "micro": "./data_new/eukaryota/",
    "archaea": "./data_new/archaea/"
}

# Fetch the path safely (defaults to current directory if theme isn't found)
path = theme_paths.get(theme, "./data_new/")

# Automatically link the theme to the corresponding organism dictionary
# (Assuming the dictionaries from Step 1 are loaded in memory)
if theme == "bacteria":
    current_orgs = bacteria
elif theme == "eukaryota":
    current_orgs = eukaryotes
elif theme == "eukaryota_core":
    current_orgs = eukaryotes_core
elif theme == "micro":
    current_orgs = microsporidia
elif theme == "archaea":
    current_orgs = archaea
else:
    raise ValueError("Invalid theme selected! Please check your spelling.")

# Extract the IDs so they are ready to be passed into your functions
org_id = list(current_orgs.keys())

print(f"Theme selected: {theme}")
print(f"Data path set to: {path}")
print(f"Loaded {len(org_id)} organisms for analysis.")
```

```python
##############################
# Generating main dictionary #
##############################

# Because we linked the 'theme' to 'current_orgs' in the previous step,
# we only need to build the dictionary for the domain we are actively analyzing!

org_dict = get_dictionary(current_orgs)

# (Optional) If you want to explicitly remind yourself what org_id is in this cell:
org_id = list(current_orgs.keys())

print(f"Main dictionary generated for {len(org_id)} organisms in the '{theme}' group.")
```

```python
# Read occupancy max path for the selected group
read_occupancy_path(path, org_dict, org_id, short=False)

print(f"Successfully loaded occupancy paths for {len(org_id)} organisms.")
```

```python
# Interpolation with the spline function

# Set the cutoff parameter (NN) based on the domain.
# Shorter tunnels (bacteria, chloroplasts, mitochondria) exit earlier at ~20 residues.
# Others (eukaryotes, archaea) exit around 25 residues.
if theme in ["bacteria", "chloroplast", "mitochondria"]:
    nn_cutoff = 20
else:
    nn_cutoff = 25

# Calculate splines using the appropriate cutoff
calculate_spline(org_dict, org_id, s=60, NN=nn_cutoff, short=False)

print(f"Splines calculated with NN cutoff = {nn_cutoff}.")
```

```python
# Write occ path avg, its interpolation, and spline into PDB files
write_paths(path, org_dict, org_id, NN=nn_cutoff)

print("Averaged occupancy paths and splines successfully saved to PDB files.")
```
